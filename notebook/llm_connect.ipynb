```python
import ollama;

response = ollama.chat(
    model='llama3.1:8B', 
    messages=[
        {
            'role': 'user',
            'content': 'Why is the sky blue?',
        },
    ],
    stream=True  # Enables real-time streaming output
)

# Print chunks as they arrive
for chunk in response:
    print(chunk['message']['content'], end='', flush=True)
```
